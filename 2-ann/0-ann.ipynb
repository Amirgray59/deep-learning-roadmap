### Backprop 

```python
def loss_function(y, y_pred) : 
    return (y - y_pred) ** 2 
```

```python
X = tt.Tensor([1,2,3,4]) 
y = tt.Tensor([2,4,6,8])
```

```python
w = 0.5
```

```python
# Learning rate
lr = 0.01

# Training
epochs = 100
```

```python

for epoch in range(epochs):

    y_pred = w * X

    loss = loss_function(y, y_pred)

    dloss_dw = np.mean(2 * (y_pred - y) * X)

    w = w - lr * dloss_dw

    if epoch % 10 == 0:
        print(
            f"Epoch: {epoch:3d} | "
            f"Loss: {loss:.6f} | "
            f"Gradient: {dloss_dw:.6f} | "
            f"Weight: {w:.6f}"
        )

print("\nFinal weight:", w)
print("Final prediction:", w * 3)
```
