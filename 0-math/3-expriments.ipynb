```python
import torch as tt 
import numpy as np
import matplotlib.pyplot as plt
```

### Functions

```python
def f(x) : 
    return np.sin(x) + 0.3 * np.sin(3*x) + 0.1 * x 

def df(x) : 
    return np.cos(x) + .9 * np.cos(3*x) + .1 
```

```python
x = np.linspace(-10,10,50)
```

```python
fx = f(x) 
dx = df(x) 
```

```python
plt.figure(figsize=(5,4)) 
plt.plot(x,fx)
plt.show()
```

```python
Epoch = 500
Lr = 0.01
```

```python
track = np.zeros((8, Epoch)) 
```

### setup 8 random starting points 

```python
track.shape
```

```python
for i in track : 
    localmin = np.random.choice(x, 1) 

    for j in range(Epoch) : 

        grad = df(localmin) 
        localmin -= Lr * grad 

        i[j] = localmin[0]
        
```

```python
track 
```

```python
plt.figure(figsize=(5,4)) 

for i in track : 
    plt.plot(x,fx) 
    plt.scatter(
        i[0] , f(i[0]) , c="b"
    )
```

```python
plt.figure(figsize=(5,4)) 

for i in track : 
    plt.plot(x,fx) 
    plt.scatter(
        i[-1] , f(i[-1]) , c="red"
    )
```

### Flexible Learning rate

```python
Epoch = 100 
Lr = .01
```

```python
for i in track : 
    localmin = np.random.choice(x, 1) 

    for j in range(Epoch) : 

        grad = df(localmin) 

        learning_rate = j * Lr 
        
        localmin -= learning_rate * grad 

        i[j] = localmin[0]
        
```

```python
plt.figure(figsize=(5,4)) 

for i in track : 
    plt.plot(x,fx) 
    plt.scatter(
        i[0] , f(i[0]) , c="b"
    )
```

```python
plt.figure(figsize=(5,4)) 

for i in track : 
    plt.plot(x,fx) 
    plt.scatter(
        i[-1] , f(i[-1]) , c="red"
    )
```
