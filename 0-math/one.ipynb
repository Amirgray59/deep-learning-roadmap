```python
# transpose 

a = torch.Tensor([
    [12,3,24,8]
])

print(a)

print("-----")

print(a.T)
```

```python
# dot product -> multiplie & sum of all values in a metrics 


m1 = np.array(
    [1,2,3,4]
)

m2 = np.array(
    [2,4,5,8]
)


np.dot(m1,m2)
```

```python
# metrix multipication 


one = np.array([
    [1,2,3,4],
    [2,4,5,3], 
    [8,3,2,4]
])

two = np.array([
    [2,4, 8],
    [3,3,3],
    [2,4,4],
    [5,8,9]
])


# one -> 3 x 4 
# two -> 4 x 3 

# three -> 3 x 3 

three = one @ two 

print(three)
```
